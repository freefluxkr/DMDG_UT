# 데이터 확인


```python
import pandas as pd


# 파일 경로 지정
file_path ="/content/bc_card_out2020_03.txt"

# with 문으로 파일 객체를 안전하게 오픈
try:
  with open(file_path, 'r', encoding='utf-8') as f:
    df = pd.read_csv(f,sep='\t',low_memory=False)

  print(df.head())

except Exception as e:
      print(f" ▲ 오류 발생: {e}")
```

# 서울특별시 데이터 상위 10개만 추출

```python
df_seoul = df[df["MEGA_CTY_NM"] == '서울특별시']
df_seoul.head(10)
```

# 서울특별시 거주/비거주 고객 수 구하기


```python
resident = df_seoul[df_seoul['CSTMR_MEGA_CTY_NM'] == '서울특별시']['CNT'].sum()

non_resident = df_seoul[df_seoul['CSTMR_MEGA_CTY_NM'] != '서울특별시']['CNT'].sum()

print(f"거주 고객 수(서울 거주): {resident:,}명")
print(f"비거주 고객 수(서울 외 거주): {non_resident:,}명")
```

# 서울특별시 거주 고객 총 소비액

```python
resident_moum = df_seoul[df_seoul['CSTMR_MEGA_CTY_NM'] == '서울특별시']['AMT'].sum()

print("=== 서울특별시 거주 고객의 총 소비액 ===\n")
print(f"총 소비액: {resident_moum:,}원")
print(f"총 소비액(조 단위): {resident_moum/1000000000000:.2f}조원")
```

# 서울특별시 거주 성별에 따른 소비액


```python
print("=== 서울특별시 거주 고객의 성별 소비액 ===\n")


spending_by_gender = df_seoul.groupby('SEX_CTGO_CD')['AMT'].sum()

print(spending_by_gender)
```

# 편의점 상위 10개 데이터 확인

```python
df_convenience = df[df['TP_BUZ_NM'] == '편 의 점']

print(df_convenience.head(10))
```

# 강남구 편의점 소비액 구하기

```python
print("=== 강남구 편의점 소비액 ===\n")

df_gangnam_convenience = df[(df['CTY_RGN_NM'] == '강남구') &
                             (df['TP_BUZ_NM'] == '편 의 점')]

print(f"강남구 편의점 데이터: {len(df_gangnam_convenience):,}개")

total_spending = df_gangnam_convenience['AMT'].sum()

print(f"\n강남구 편의점 총 소비액: {total_spending:,}원")
print(f"강남구 편의점 총 소비액(억 단위): {total_spending/100000000:.2f}억원")

total_count = df_gangnam_convenience['CNT'].sum()
print(f"강남구 편의점 총 거래 건수: {total_count:,}건")
```

# 서울특별시 거주 고객 총 소비액, 비거주 고객 총 소비액 구하기

```python
resident_spending = df_seoul[df_seoul['CSTMR_MEGA_CTY_NM'] == '서울특별시']['AMT'].sum()
resident_count = df_seoul[df_seoul['CSTMR_MEGA_CTY_NM'] == '서울특별시']['CNT'].sum()


non_resident_spending = df_seoul[df_seoul['CSTMR_MEGA_CTY_NM'] != '서울특별시']['AMT'].sum()
non_resident_count = df_seoul[df_seoul['CSTMR_MEGA_CTY_NM'] != '서울특별시']['CNT'].sum()

print("=== 거주 고객(서울 거주) ===")
print(f"총 소비액: {resident_spending:,}원")
print(f"총 소비액(억 단위): {resident_spending/100000000:.2f}억원")



print("\n=== 비거주 고객(서울 외 거주) ===")
print(f"총 소비액: {non_resident_spending:,}원")
print(f"총 소비액(억 단위): {non_resident_spending/100000000:.2f}억원")

```

# 거주지 소재 편의점 소비액 구하기

```python
df_convenience = df[df['TP_BUZ_NM'] == '편 의 점']

print(f"전체 편의점 데이터: {len(df_convenience):,}개\n")

df_home_convenience = df_convenience[(df_convenience['MEGA_CTY_NM'] == df_convenience['CSTMR_MEGA_CTY_NM']) &
                                      (df_convenience['CTY_RGN_NM'] == df_convenience['CSTMR_CTY_RGN_NM'])]

df_other_convenience = df_convenience[(df_convenience['MEGA_CTY_NM'] != df_convenience['CSTMR_MEGA_CTY_NM']) |
                                       (df_convenience['CTY_RGN_NM'] != df_convenience['CSTMR_CTY_RGN_NM'])]

print("=== 거주지 소재 편의점 ===")
home_spending = df_home_convenience['AMT'].sum()
home_count = df_home_convenience['CNT'].sum()
print(f"총 소비액(억 단위): {home_spending/100000000:.2f}억원")


print("\n=== 비거주지 소재 편의점 ===")
other_spending = df_other_convenience['AMT'].sum()
other_count = df_other_convenience['CNT'].sum()
print(f"총 소비액(억 단위): {other_spending/100000000:.2f}억원")
```
