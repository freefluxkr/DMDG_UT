# **# 7. 서울시 거주/비거주 고객의 편의점 소비액**

```python
편의점_거주별_소비액 = 편의점.groupby("거주구분")["AMT"].sum()

for 구분, 금액 in 편의점_거주별_소비액.items():
    print(구분, f"{금액:,}원")
```

# **# 8. 거주지 소재 편의점 소비액**

고객이 사는 지역과 편의점이 있는 지역이 같은 소비만 고르기



*  고객 거주지: 서울특별시 강남구
*  편의점 위치: 서울특별시 강남구




```python
거주지_편의점 = 편의점[
    (편의점["MEGA_CTY_NM"] == 편의점["CSTMR_MEGA_CTY_NM"]) &
    (편의점["CTY_RGN_NM"] == 편의점["CSTMR_CTY_RGN_NM"])
]

거주지_편의점_소비액 = 거주지_편의점["AMT"].sum()

print(f"{거주지_편의점_소비액:,}원")
```

# **# 10. 거주구분별 건당 소비금액**

```python
건당_소비금액 = df.groupby("거주구분").agg(
    총소비액=("AMT", "sum"),
    이용건수=("CNT", "sum")
)

건당_소비금액["건당소비액"] = (
    건당_소비금액["총소비액"]
    / 건당_소비금액["이용건수"]
)

건당_소비금액
```

# **결론**

서울시 카드 소비 데이터를 분석한 결과, 서울 거주 고객은 서울 비거주
고객보다 카드 이용건수가 많았다. 반면 총소비액은 서울 비거주 고객이
더 높게 나타났다.


이용건수와 총소비액을 함께 비교한 결과, 서울 비거주 고객의 건당
소비금액이 서울 거주 고객보다 높게 나타났다. 따라서 서울 거주 고객은
상대적으로 자주 카드를 이용하고, 서울 비거주 고객은 한 번 이용할 때
더 큰 금액을 소비하는 경향이 있다고 해석할 수 있다.

편의점 소비액은 총 79,987,167,291원이었으며, 이 중 서울 거주 고객의
소비액은 66,513,977,146원으로 서울 비거주 고객보다 높게 나타났다.
이를 통해 서울 지역 편의점 소비는 주로 서울 거주 고객의 일상적인
생활 소비에서 발생한 것으로 볼 수 있다.

강남구 편의점 소비액은 8,170,947,461원이었으며, 고객의 거주지와
편의점 소재지가 같은 경우를 지역별로 분석한 결과 강남구의 소비액이
가장 높았다. 이를 통해 편의점 소비가 고객의 생활권과 밀접한 관련이
있음을 확인할 수 있었다.
