```python
# 파일불러오기

import pandas as pd

# 파일경로 지정

file_path = r"../data/bc_card_out2020_03.txt"

# with 문으로 파일 객체를 안전하게 오픈

try :

    with open(file_path, "r", encoding='utf-8') as f:

        df = pd.read_csv(f, sep='\t', low_memory=False)

    print("성공적으로 데이터를 불러왔습니다.")

    print(df.head())

except Exception as e:

    print(f"오류발생: {e}")

# 데이터를 확인해보자

print("데이터 크기 : " , df.shape)

# 결측치 확인해보기

df.info()
```

```python
df_clean = df[[
    'CSTMR_MEGA_CTY_NM',
    'SEX_CTGO_CD',
    'AMT',
    'TP_BUZ_NM',
    'CTY_RGN_NM',
    'CSTMR_CTY_RGN_NM']].copy()
print(df_clean.head())
```

```python
gender_map = {1: '남성', 2: '여성'}
df_clean['SEX_CTGO_CD'] = df_clean['SEX_CTGO_CD'].map(gender_map)
print(df_clean.head())
```

```python
city_counts = df_clean['CSTMR_MEGA_CTY_NM'].value_counts()
print(city_counts)
```

```python
seoul_users = df_clean[df_clean['CSTMR_MEGA_CTY_NM'] == '서울특별시'].shape[0]
non_seoul_users = df_clean[df_clean['CSTMR_MEGA_CTY_NM'] != '서울특별시'].shape[0]
print("① 서울시 거주 고객 수 :", seoul_users, "명")
print("① 비거주(서울 외) 고객 수 :", non_seoul_users, "명")
```

```python
total_amt = df_clean['AMT'].sum()
print("② 전체 고객의 총 소비액 :", total_amt, "원")
print("② 만원 단위로 보기 : 약", round(total_amt / 10000), "만원")
```

```python
gender_amt = df_clean.groupby('SEX_CTGO_CD')['AMT'].sum()
print(gender_amt)
print("\n[만원 단위로 보기]")
print(round(gender_amt / 10000))
```

```python
df_store = df_clean[df_clean['TP_BUZ_NM'] == '편 의 점']
print("편의점 데이터 개수 :", df_store.shape[0], "건")
print(df_store.head())
```

```python
convenience_total_amt = df_store['AMT'].sum()
print("① 편의점 총 소비액 :", convenience_total_amt, "원")
print("① 만원 단위로 보기 : 약", round(convenience_total_amt / 10000), "만원")
```

```python
df_gangnam_store = df_store[df_store['CTY_RGN_NM'] == '강남구']
gangnam_store_amt = df_gangnam_store['AMT'].sum()
print("② 강남구 편의점 데이터 개수 :", df_gangnam_store.shape[0], "건")
print("② 강남구 편의점 총 소비액 :", gangnam_store_amt, "원")
print("② 만원 단위로 보기 : 약", round(gangnam_store_amt / 10000), "만원")
```

```python
city_amt = df_clean.groupby('CSTMR_MEGA_CTY_NM')['AMT'].sum()
print(city_amt)
print("\n[만원 단위로 보기]")
print(round(city_amt / 10000))
```

```python
seoul_amt = df_clean[df_clean['CSTMR_MEGA_CTY_NM'] == '서울특별시']['AMT'].sum()
non_seoul_amt = df_clean[df_clean['CSTMR_MEGA_CTY_NM'] != '서울특별시']['AMT'].sum()
print("③ 서울시 거주 고객 총 소비액 :", seoul_amt, "원")
print("③ 서울시 거주 고객 소비액 (만원 단위) : 약", round(seoul_amt / 10000), "만원")
print("-" * 40)
print("③ 비거주(서울 외) 고객 총 소비액 :", non_seoul_amt, "원")
print("③ 비거주(서울 외) 고객 소비액 (만원 단위) : 약", round(non_seoul_amt / 10000), "만원")
```

```python
df_home_store = df_store[df_store['CTY_RGN_NM'] == df_store['CSTMR_CTY_RGN_NM']]
home_store_amt = df_home_store['AMT'].sum()
print("④ 거주지 소재 편의점 결제 건수 :", df_home_store.shape[0], "건")
print("④ 거주지 소재 편의점 총 소비액 :", home_store_amt, "원")
print("④ 만원 단위로 보기 : 약", round(home_store_amt / 10000), "만원")
```
