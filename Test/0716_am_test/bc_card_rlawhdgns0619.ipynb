```python
import pandas as pd


df = pd.read_csv('bc_card_out2020_03.txt', sep='\t', encoding='utf-8')

df.isnull().sum()

selected_cols = ['MEGA_CTY_NM','CSTMR_MEGA_CTY_NM','CTY_RGN_NM','SEX_CTGO_CD', 'TP_BUZ_NM', 'AMT']

df_selected = df[selected_cols].copy()


df_selected['usage_type'] = df_selected.apply(
    lambda x: '거주고객소비액' if x['CSTMR_MEGA_CTY_NM'] == x['MEGA_CTY_NM'] else '비거주고객소비액',
    axis=1
)


result = df_selected.groupby('usage_type')['AMT'].sum()

total_amount = result.sum()

print("--- 유형별 소비액 ---")
print(result)
print("-" * 20)
print(f"전체 소비액: {total_amount:,} 원")
```

```python
gender_summary = df_selected.groupby('SEX_CTGO_CD')['AMT'].sum()


gender_summary.index = gender_summary.index.map({1: '남성', 2: '여성'})


print("--- 성별 소비액 합계 ---")
print(gender_summary)
```

```python
df_conveni = df_selected[df_selected['TP_BUZ_NM'].str.contains('편 의 점', na=False)].copy()


df_gangnam_conveni = df_conveni[df_conveni['CTY_RGN_NM'] == '강남구'].copy()


total_conveni_amt = df_conveni['AMT'].sum()
gangnam_conveni_amt = df_gangnam_conveni['AMT'].sum()

print("--- 편의점 소비액 분석 ---")
print(f"전체 편의점 소비액: {total_conveni_amt:,} 원")
print(f"강남구 편의점 소비액: {gangnam_conveni_amt:,} 원")

```
