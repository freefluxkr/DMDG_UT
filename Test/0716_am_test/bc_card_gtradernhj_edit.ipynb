```python
'''
-------------------------------------------------------------
### 0716 오전시험 데이터 전처리 작업계획
-------------------------------------------------------------
1. 데이터 소개. > 세개의 데이터파일을 불러와서 하나의df파일로 합치기.
    - 데이터 크기, 컬럼명, 데이터 타입 확인후에 진행
    - 사용할 함수 (df.shape(), df.info(), df.head())

2. 데이터 전처리
    - 결측치 처리: df.isnull().sunm()
    - 컬럼명 정리.
    - 중복제거.

3. 데이터 분석
    - 서울시 거주/비거주 고객분석
      - 고객 수 : df['주소'].str.contains('서울').value_counts
      - 총 소비액 : df['소비액'].sum()
      - 성비 소비액 : df.groupby('성별')['소비액'].sum()
    - 편의점 소비 분석
      - 전체 편의점 소비액 : df[df['업종']=='편의점']['소비액'].sum()
      - 강남구
    - 지역별 소비 분석
      - 서울시 거주/비거주 소비액: df.groupby('서울시거주여부')['소비액'].sum()
      - 거주지 지역별 편의점 소비액: df[df['업종']=='편의점'].groupby('주소')['소비액'].sum()
'''
```

```python
import pandas as pd

# 파일경로 지정

file_path = r"../data/bc_card_out2020_03.txt"

# with 문으로 파일 객체를 안전하게 오픈

try :

    with open(file_path, "r", encoding='utf-8') as f:

        df = pd.read_csv(f, sep='\t', low_memory=False)

    print("성공적으로 데이터를 불러왔습니다.")

    print(df.head())

except Exception as e:

    print(f"오류발생: {e}")

# 데이터를 확인해보자

print("데이터 크기 : " , df.shape)

# 결측치 확인해보기

df.info()
```

```python
df.head() #필요한 컬럼 정보 확인.
# 서울시 거주 비거주 인구수 구하기.
# 필요한 컬럼 MEGA_CITY_NO,
# MEGA CITY NUMBER = 11인 AMT의 합과, 나머지의합
```

```python
df.isnull().sum() # 결측치 확
```

```python
df_clean = df.dropna() #결측치가 비교적 적으므로 dropna함수로 삭
```

```python
# 서울특별시 도시넘버 = 11
df_clean['서울거주여부'] = df_clean['MEGA_CTY_NO'].apply(lambda x: '서울거주' if x == 11 else '비거주')

# 집계 결과 확인
print(df_clean['서울거주여부'].value_counts())
```

```python
#서울시 인구를 컬럼으로 추
df_clean.loc[:, '서울거주여부'] = df_clean['MEGA_CTY_NO'].apply(lambda x: '서울거주' if x == 11 else '비거주')
```

```python
# 총 소비액 계산
total_amt = df_clean['AMT'].sum()
print("총 소비액:", total_amt)
```

```python
#성별소비액 구하기. 사용할 컬럼정보는 'sex_ctgo_cd', 'amt'
# 성별 코드 → 라벨 매핑
gender_map = {1: '남성', 2: '여성'}

df_clean['성별'] = df_clean['SEX_CTGO_CD'].map(gender_map)

gender_amt = df_clean.groupby('성별')['AMT'].sum()
print(gender_amt)
```

```python
#편의점 소비액 구하기, MAIN_BUZ_DESC > 컬럼에 내용들을 확인
# 업종명 고유값 확인
unique_buz = df_clean['MAIN_BUZ_DESC'].unique()
print(unique_buz)

# 업종명 개수 확인
print("업종명 종류 수:", df_clean['MAIN_BUZ_DESC'].nunique())
```

```python
#업종명 '생활'로 구분해서 그 합을 구하기
# 업종명이 '생활'인 데이터만 필터링 후 소비액 합계
life_amt = df_clean.loc[df_clean['MAIN_BUZ_DESC'] == '생활', 'AMT'].sum()
print("생활 업종 총 소비액:", life_amt)
```

```python
#강남구 편의점 소비액 총합 구하기, CSTMR_CTY_RGN_NM > 강남구로 sort
gangnam_df = df_clean.loc[df_clean['CSTMR_CTY_RGN_NM'] == '강남구']
print(gangnam_df.head())  # 강남구 데이터 샘플 확인
```

```python
#강남구 업종별 집계내역
gangnam_buz_amt = gangnam_df.groupby('MAIN_BUZ_DESC')['AMT'].sum()
print(gangnam_buz_amt)
```

```python
gangnam_convenience_amt = gangnam_df.loc[gangnam_df['MAIN_BUZ_DESC'] == '생활', 'AMT'].sum()
print("강남구 편의점 총 소비액:", gangnam_convenience_amt)
```

```python
#서울시 거주 비거주 기준 고객의 소비액 구하기
#서울 거주자 '생활' 소비
seoul_life_amt = df_clean.loc[
    (df_clean['MEGA_CTY_NO'] == 11) & (df_clean['MAIN_BUZ_DESC'] == '생활'),
    'AMT'
].sum()
print("서울거주자의 생활 업종 소비액:", seoul_life_amt)

# 비거주 소비액
non_seoul_life_amt = df_clean.loc[
    (df_clean['MEGA_CTY_NO'] != 11) & (df_clean['MAIN_BUZ_DESC'] == '생활'),
    'AMT'
].sum()
print("비거주자의 생활 업종 소비액:", non_seoul_life_amt)



```

```python
#거주지 소재별 편의점 소비액 구하기
# 거주지 소재별 편의점 소비액 집계
convenience_by_region = df_clean.loc[df_clean['MAIN_BUZ_DESC'] == '생활'] \
    .groupby('CSTMR_CTY_RGN_NM')['AMT'].sum()

print(convenience_by_region)
```
